# Analýza behania a kopcovej záťaže (`Cardio Activities`)
**Google Data Analytics Capstone Project – Case Study**

* **GitHub repozitár:** [Hill-running-performance-analysis](https://github.com/petermich84-boop/Hill-running-performance-analysis)
* **Zdroj dát:** Verejný dataset *Cardio Activities* (`cardioActivities.csv`) na Kaggle

---

## 1. Business task (Ask)
* **Scenár:** Tréningový bežecký program / koučingová platforma, ktorá chce lepšie porozumieť súvislosti medzi prevýšením, dĺžkou behu, tempom a srdcovou frekvenciou pri plánovaní kopcových a intenzívnych tréningov.
* **Analytická otázka:** *Ako sa menia vzdialenosť, trvanie, tempo a priemerná srdcová frekvencia v závislosti od relatívneho prevýšenia na kilometer (`climb_per_km`) – a prečo sa v súhrnných dátach kopcové behy ukazujú skôr ako dlhšie vytrvalostné tréningy než krátke šprinty do kopca?*
* **Sledované metriky:** vzdialenosť (km), trvanie (min), priemerné tempo (min/km), priemerná rýchlosť (km/h), celkové prevýšenie (`Climb (m)`), relatívne prevýšenie na kilometer (`climb_per_km` v m/km) a priemerná srdcová frekvencia (`bpm`).

## 2. Dátové zdroje (Prepare)
* **Štruktúra a rozsah:** Po vyfiltrovaní bežeckých aktivít (`Running`) obsahuje súbor **459 behov** so súhrnnými metrikami za celú aktivitu (priemerná srdcová frekvencia je dostupná pri **277 behoch**).
* **Kredibilita, etika a obmedzenia:** Dataset neobsahuje osobné identifikačné údaje. Hlavným obmedzením je, že ide o **súhrnné hodnoty za celý beh** (nie jednotlivé šprintové opakovania do kopca po metroch) a pri časti záznamov chýba meranie tepu.
